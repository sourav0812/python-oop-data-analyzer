```python
########################## Q2, part 1: ########################## 

class Project (Student): # class named Project with a bunch of attributes that is associated with Student.
#     we are inheriting from the Student Class

    
# This initializes the Project object and uses a loop to ensure that the data points 
# inputted are only numerical. We are also handling error case here as well.
    def __init__(self, student_id, major, university, project_id, data_points):
        # this is how we know that this class is a subclass of Student
        super().__init__(student_id, major, university )
        self.project_id = project_id
        
        for value in data_points:
            if type(value) != int and type(value) != float:
                raise TypeError ( f"Please input a numerical data point. {value}" 
                                 f" is a non numerical data point.")
            
        self.data_points = data_points
        self.__analysis_results = {}
        self.__active = True

########################## Q2, part 2: ##########################

# This allows us to add new data point to data_points. it also checks if the data point that is inputted
# is a numerical data point or not. error handling is also taken into consideration.

    def add_data(self, new_data):
        if type(new_data) != int and type(new_data) != float:
            raise TypeError ( f"Please input a numerical data point. {new_data}, is a non numerical data point.")

        self.data_points.append(new_data)

        
# this is a getter method that returns the __analysis_results, a dictionary. __analysis_results is private
    def get_results(self):
        return self.__analysis_results

# this method returns the active status of Project. active status is True by default, __active is private
    def is_active(self):
        return self.__active

# sets the active status of the project. __active is private
    def set_active(self,status):
        self.__active = status
        
# returns the mean, median, and sample variance of the given data_points and then stores them in the
# dictionary __analysis_results. __analysis_results is private. if the given list in data_points is empty, 
# a message is printed (ZeroDivisionError).
        
    def perform_analysis(self):

        try:
            ########################## mean ##########################

            total_of_data_points = 0
            for each_value in self.data_points:
                total_of_data_points = total_of_data_points + each_value
            mean_of_data_points = total_of_data_points / len(self.data_points)

            ########################## median ##########################
            # making a copy of the data_points list data
            sorted_data_points = sorted(self.data_points)

            # two middle num prob, even list:
            if len(sorted_data_points) % 2 == 0:
                middleNum1 = len(sorted_data_points) // 2
                middleNum2 = ((len(sorted_data_points) // 2) - 1)
                median = float((sorted_data_points[middleNum1] + sorted_data_points[middleNum2] ) / 2)

            else: # for odd num of list
                median_index = len(sorted_data_points) // 2  # we are rounding down.
                median = float(sorted_data_points[median_index])

            ########################## Sample Variance: ##########################

            Sqrd_Total_sum = 0
            for each_value in self.data_points:
                Sqrd_Total_sum = (each_value - mean_of_data_points)**2 + Sqrd_Total_sum

            Sample_variance = Sqrd_Total_sum / (len(self.data_points) - 1)


            ########################## Storing Results:  ##########################
            self.__analysis_results ["Mean"] = mean_of_data_points
            self.__analysis_results ["Median"] = median
            self.__analysis_results ["Sample Variance"] = Sample_variance

        except ZeroDivisionError:
            print ("Data_points list is empty--Analysis cannot be performed.")

```
